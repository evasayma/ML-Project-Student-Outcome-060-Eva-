```python
import numpy as np

np.random.seed(42)

marks = np.array([55, 72, 48, 90, 66, 81])

print('Marks:', marks)
print('Passed:', marks[marks >= 50])

print('Minimum:', marks.min())
print('Standard Deviation:', marks.std())

bonus_marks = marks + 5
print('Bonus Marks:', bonus_marks)
print('Original Marks:', marks)

print('Number of Passed Students:', np.sum(marks >= 50))


matrix = marks.reshape(2, 3)
print('2 × 3 Matrix:')
print(matrix)
```

```python
import pandas as pd

data = {
    'student': ['Asha', 'Babu', 'Chayan', 'Dipa', 'Emon', 'Farah'],
    'study_hours': [2, 5, 1, 7, 3, 6],
    'sleep_hours': [8, 6, 9, 5, 7, 6],
    'attendance': [70, 88, 55, 95, 75, 91],
    'snacks': [3, 2, 5, 1, 4, 2],
    'passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

print("Original DataFrame:")
display(df)

features = ['study_hours', 'sleep_hours', 'attendance', 'snacks']
label = 'passed'

print("\nFeatures:", features)
print("Label:", label)

print("\nNumber of samples:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nStudents with attendance at least 80:")
display(df[df['attendance'] >= 80])

average_study = df[df['passed'] == 'Yes']['study_hours'].mean()

print("\n19. Average study hours of passed students:", average_study)

print("\n20. DataFrame sorted by attendance:")
display(df.sort_values(by='attendance', ascending=False))
```

```python
import random
import numpy as np

def generate_scores(use_seed):
    if use_seed:
        random.seed(42)
        np.random.seed(42)

    return np.random.randint(40, 101, size=5)


print('Seeded run 1:', generate_scores(True))
print('Seeded run 2:', generate_scores(True))

print('Unseeded run 1:', generate_scores(False))
print('Unseeded run 2:', generate_scores(False))
```

```python
import random
import numpy as np
def generate_scores(use_seed):
    if use_seed:
      random.seed(42)
      np.random.seed(42)
    return np.random.randint(40, 101, size=5)
print('Seeded run 1:', generate_scores(True))
print('Seeded run 2:', generate_scores(True))
print('Unseeded run 1:', generate_scores(False))
print('Unseeded run 2:', generate_scores(False))
print('The seed problem in machine learning occurs because random processes such as splitting data into training and testing sets,')
print('can produce different results each time. If we use a fixed random seed,such as 42, the same random results are generated every time,making the experiment reproducible.')
print('Without a seed,different runs may produce different results and accuracy scores.')
print('Therefore,using the same seed helps researchers compare models fairly and ensures that their results can be repeated.')
```
